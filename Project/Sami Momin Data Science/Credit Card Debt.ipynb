```python
df = pd.read_csv("credit_card_debt.csv")
```

```python
print(df.shape)
```

```python
missing = pd.DataFrame({
    "missing_count": df.isna().sum(),
    "missing_pct": (df.isna().mean() * 100).round(2),
})

missing - missing[missing["missing_count"] > 0].sort_values(
    "missing_pct", ascending=False
)

print(missing)
```

```python
duplicate_cols = []

cols = df.columns

for i, col1 in enumerate(cols):
    for col2 in cols[i+1:]:
        if df[col1].equals(df[col2]):
            duplicate_cols.append(col2)

print("Duplicate columns: ", duplicate_cols)
```

```python
full_dupes = df[df.duplicated(keep = False)]

print(f"Rows that are part of a duplicate group: {len(full_dupes)}")
print(f"Number of duplicate rows beyond the first: {df.duplicated().sum()}")
```
